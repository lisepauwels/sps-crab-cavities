```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_csv('impacts_0.10.1.csv')
df2 = pd.read_csv('impacts_0.6.8.csv')
```

# Interaction types

```python
interaction_types, counts = np.unique(np.array(df.interaction_type), return_counts=True)
print(interaction_types)
print(counts)
```

```python
interaction_types2, counts2 = np.unique(np.array(df2.interaction_type), return_counts=True)
print(interaction_types2)
print(counts2)
```

```python
pos = np.arange(len(interaction_types))  # assumes both lists have same labels in same order
width = 0.35

fig, ax = plt.subplots()
ax.bar(pos - width/2, counts, width, label='xcoll 0.10.1')
ax.bar(pos + width/2, counts2, width, color='orange', label='xcoll 0.6.8')

ax.set_ylabel('Counts')
ax.set_title('Counts of interaction types')
ax.set_xticks(pos)
ax.set_xticklabels(interaction_types, rotation=90)
ax.legend()
plt.tight_layout()
plt.show()
```

# Distribution of particles

```python
import json
with open('monitor_0.10.1.json', 'r') as f:
    monitor_dict = json.load(f)
```

```python
with open('monitor_0.6.8.json', 'r') as f:
    monitor_dict2 = json.load(f)
```

```python
x = np.array(monitor_dict['data']['x']).reshape(10000, 3000)
y = np.array(monitor_dict['data']['y']).reshape(10000, 3000)
px = np.array(monitor_dict['data']['px']).reshape(10000, 3000)
py = np.array(monitor_dict['data']['py']).reshape(10000, 3000)
zeta = np.array(monitor_dict['data']['zeta']).reshape(10000, 3000)
delta = np.array(monitor_dict['data']['delta']).reshape(10000, 3000)
```

```python
x2 = np.array(monitor_dict2['data']['x']).reshape(10000, 3000)
y2 = np.array(monitor_dict2['data']['y']).reshape(10000, 3000)
px2 = np.array(monitor_dict2['data']['px']).reshape(10000, 3000)
py2 = np.array(monitor_dict2['data']['py']).reshape(10000, 3000)
zeta2 = np.array(monitor_dict2['data']['zeta']).reshape(10000, 3000)
delta2 = np.array(monitor_dict2['data']['delta']).reshape(10000, 3000)
```

```python
#[(df.turn > 500) & (df.turn < 1500)]
cmap = plt.get_cmap('viridis')
norm = plt.Normalize(vmin=0, vmax=3000)
fig, ax = plt.subplots(1,2, figsize=(12,6))
for turn in df.turn.unique():
    ax[0].scatter(x[:, turn], y[:, turn], color=cmap(norm(turn)))
    ax[1].scatter(x2[:, turn], y2[:, turn], color=cmap(norm(turn)))

ax[0].grid()
limsx0 = ax[0].get_xlim()
limsx1 = ax[1].get_xlim()
limxs = [min(limsx0[0], limsx1[0]), max(limsx0[1], limsx1[1])]
limsy0 = ax[0].get_ylim()
limsy1 = ax[1].get_ylim()
limys = [min(limsy0[0], limsy1[0]), max(limsy0[1], limsy1[1])]
ax[0].set_xlim(limxs)
ax[0].set_title('xcoll 0.10.1')
ax[1].grid()
ax[1].set_xlim(limxs)
ax[1].set_title('xcoll 0.6.8')
ax[0].set_ylim(limys)
ax[1].set_ylim(limys)
plt.tight_layout()
plt.show()
```
